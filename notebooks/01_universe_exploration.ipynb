# 01 · Universe Exploration

This notebook walks through:
1. Loading the pre-built watchlist CSV
2. Visualising RS rank distribution
3. Inspecting ADR% and dollar volume filters
4. Reviewing sector / theme breakdown

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

WATCHLIST_PATH = Path('../data/outputs/watchlist.csv')
df = pd.read_csv(WATCHLIST_PATH) if WATCHLIST_PATH.exists() else pd.DataFrame()
print(f'Loaded {len(df)} symbols')
df.head()
```

## RS Rank Distribution

```python
if not df.empty and 'rs_rank_pct' in df.columns:
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.hist(df['rs_rank_pct'], bins=20, edgecolor='black')
    ax.set_xlabel('RS Rank Percentile')
    ax.set_ylabel('Count')
    ax.set_title('Relative Strength Rank Distribution')
    plt.tight_layout()
    plt.show()
```

## ADR% Distribution

```python
if not df.empty and 'adr_pct' in df.columns:
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.hist(df['adr_pct'], bins=20, color='orange', edgecolor='black')
    ax.set_xlabel('ADR %')
    ax.set_ylabel('Count')
    ax.set_title('Average Daily Range Distribution')
    plt.tight_layout()
    plt.show()
```

## Sector Breakdown

```python
if not df.empty and 'sector' in df.columns:
    sector_counts = df['sector'].value_counts()
    fig, ax = plt.subplots(figsize=(10, 4))
    sector_counts.plot(kind='barh', ax=ax, color='steelblue')
    ax.set_xlabel('Count')
    ax.set_title('Watchlist Sector Breakdown')
    plt.tight_layout()
    plt.show()
```

## Top 20 by RS Rank

```python
if not df.empty:
    cols = [c for c in ['symbol','price','rs_rank_pct','adr_pct','avg_dollar_volume_20d','sector','theme'] if c in df.columns]
    df.sort_values('rs_rank_pct', ascending=False).head(20)[cols]
```
